## 1. The CNN outputs one Q-value per action

The network receives `(4, 42, 42)` and returns `(2,)`: one value for `do nothing` and one for `flap`.

```python
import sys
from pathlib import Path

import numpy as np
import torch

sys.path.append(str(Path('03_deep_q_learning').resolve()))
from train_deep_q_network import QNetwork, OBSERVATION_SHAPE

network = QNetwork()
dummy_stack = torch.zeros((1, *OBSERVATION_SHAPE))
print('input shape:', tuple(dummy_stack.shape))
print('Q-value shape:', tuple(network(dummy_stack).shape))
print('example Q-values:', network(dummy_stack).detach().numpy()[0])
```

### Checkpoint 1 — action values

If the network outputs `[1.4, 2.1]`, it selects `flap` greedily because `2.1` is larger.
